```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline 
import seaborn as sns
```

```python
df=pd.read_csv(r'C:\Users\sanjoy\Documents\jupyter\Diwali Sales Data.csv',encoding ="unicode_escape")
```

```python
------------------------------------------------------------------𝐂𝐋𝐄𝐀𝐍𝐈𝐍𝐆---------------------------------------------------------------------------
```

```python
df.drop(['Status','unnamed1'], axis=1, inplace=True)
```

```python
pd.isnull(df).sum()
```

```python
df.dropna(inplace=True)
```

```python
df['Amount'] = df['Amount'].astype('int')
```

```python
*****************************************************************𝗗𝗮𝘁𝗮 𝗮𝗻𝗮𝗹𝘆𝘀𝗶𝘀**************************************************************************
```

```python
-----------------------------------------------------------------𝐆𝐄𝐍𝐃𝐄𝐑-------------------------------------------------------------------------------
```

```python
df.columns
```

```python
ax=sns.countplot(x='Gender',data = df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_gn= df.groupby(['Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Gender',y='Amount',data =sale_gn)
```

```python
#From this graphs we can conclude that most of the buyer is female and they have more purchasing power than male.
```

```python
-----------------------------------------------------------------𝐀𝐆𝐄----------------------------------------------------------------------------------
```

```python
df.columns
```

```python
ax=sns.countplot(x='Age Group',hue = 'Gender',data=df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_agn= df.groupby(['Age Group'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.barplot(x='Age Group',y='Amount',data =sale_agn)
```

```python
#The conclusion arise that most of the buyer is the age group of 26-35 yrs and in gender is female.
```

```python
-------------------------------------------------------------------𝐒𝐓𝐀𝐓𝐄------------------------------------------------------------------------------
```

```python
df.columns
```

```python
sale_state= df.groupby(['State'],as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,10)})
sns.barplot(x='State',y='Orders',data =sale_state)
```

```python
sale_state2= df.groupby(['State'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.barplot(x='State',y='Amount',data =sale_state2)
```

```python
#the conclusion arise from above that the most of orders & total ammount/sale are from Uttar Pradesh,Maharashtra and karnataka respectively.
```

```python
-------------------------------------------------------------------𝐌𝐚𝐫𝐢𝐭𝐚𝐥 𝐒𝐭𝐚𝐭𝐮𝐬-----------------------------------------------------------------------
```

```python
df.columns
```

```python
ax=sns.countplot(x='Marital_Status',data=df)

sns.set(rc={'figure.figsize':(10,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_mar= df.groupby(['Marital_Status','Gender'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(10,5)})
sns.barplot(x='Marital_Status',y='Amount',data =sale_mar,hue='Gender')
```

```python
#From above we can conclude that most of the buyer are married(female) and they have high purchasing power.
```

```python
-------------------------------------------------------------------𝐎𝐂𝐂𝐔𝐏𝐀𝐓𝐈𝐎𝐍------------------------------------------------------------------------
```

```python
sns.set(rc={'figure.figsize':(20,10)})
ax=sns.countplot(x='Occupation',data=df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_ocu= df.groupby(['Occupation'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.set(rc={'figure.figsize':(20,10)})
sns.barplot(x='Occupation',y='Amount',data =sale_ocu)
```

```python
#From the above graph we can conclude that most buyer are working in IT,Aviation and Healthcare sector
```

```python
--------------------------------------------------------------𝐏𝐑𝐎𝐃𝐔𝐂𝐓 𝐂𝐀𝐓𝐄𝐆𝐎𝐑𝐘----------------------------------------------------------------------
```

```python
sns.set(rc={'figure.figsize':(25,10)})
ax=sns.countplot(x='Product_Category',data=df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sale_pro= df.groupby(['Product_Category'],as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)
sns.set(rc={'figure.figsize':(20,10)})
sns.barplot(x='Product_Category',y='Amount',data =sale_pro)
```

```python
#from the above we can conclude that most sold products are from Food,Clothing and Electronic category
```
